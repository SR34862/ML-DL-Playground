```python
from sklearn.model_selection import train_test_split
```

```python
a = df.drop("Label", axis=1)
```

```python
x = a.drop("Freq_1", axis=1)
y = a["Freq_2"]
```

```python
x_column = x.iloc[:, [0]]
threshold = y.median()
y_binary = (y > threshold).astype(int)
x, y_binary = x.align(y_binary, join='inner', axis=0)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_column, y_binary, test_size=0.1, random_state=325)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
scaler.fit(x_train)
scaler.fit_transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
l1 = LinearRegression()
```

```python
l1.fit(x_train, y_train)
```

```python
y_pred = l1.predict(x_column)
```

```python
plt.scatter(x_column, y)
plt.plot(x_column, y_pred)
plt.show()
```
