```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data/cement_slump.csv")
```

```python
x = df.drop("Water", axis=1)
y = df.drop("Compressive Strength (28-day)(Mpa)", axis=1)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.1, random_state=325)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
scaler.fit_transform(x_train)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
l1 = LogisticRegression(solver="saga", multi_class="ovr", max_iter=5000)
```

```python
threshold = y_train.median()
y_train_binary = (y_train > threshold).astype(int)
```

```python
a = y_train_binary.drop("Cement", axis=1)
b = a.drop("Slag", axis=1)
c = b.drop("Fly ash", axis=1)
d = c.drop("Water", axis=1)
e = d.drop("SP", axis=1)
f = e.drop("Coarse Aggr.", axis=1)
g = f.drop("Fine Aggr.", axis=1)
h = g.drop("SLUMP(cm)", axis=1)
i = h.drop("FLOW(cm)", axis=1)
```

```python
l1.fit(x_train, h)
```

```python
y_train_binary.columns
```

```python
y_pred = l1.predict(x_train)
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import r2_score
```

```python
threshold = y_train.median()
y_binary = (y_train > threshold).astype(int)
pred_threshold = np.median(y_pred)
pred_binary = (y_pred > pred_threshold).astype(int)
score = accuracy_score(y_binary, pred_binary)
```
