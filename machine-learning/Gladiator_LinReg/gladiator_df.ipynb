```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data/gladiator_data.csv")
```

```python
sns.barplot(x=df["Patron Wealth"], y=df["Wins"].value_counts())
```

```python
sns.barplot(x=df["Weapon of Choice"], y=df["Wins"].value_counts())
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
```

```python
x = df[["Battle Experience"]]
y= df[["Wins"]]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.1, random_state=325)
```

```python
L1 = LinearRegression()
L1.fit(x_train, y_train)
```

```python
y_pred = L1.predict(x)
```

```python
plt.plot(x_test, y_pred)
```
