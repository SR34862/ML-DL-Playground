```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("data/Ames_outliers_removed.csv")
```

```python
df.head()
```

```python
sns.scatterplot(data=df, x=df["Overall Qual"], y=df["SalePrice"])
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
L1 = LogisticRegression()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler 
```

```python
x = df[["Overall Qual"]]
y = df[["SalePrice"]]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, random_state=325, test_size=0.1)
```

```python
scaler = StandardScaler()
```

```python
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
L1 = LogisticRegression(max_iter=1000)
```

```python
L1.fit(x_train, y_train)
```

```python
y_pred = L1.predict(x_test)
```

```python
plt.figure(figsize=(10, 7))
sns.scatterplot(x=x_train, y=y_pred)
plt.show()
```
