```python
df = pd.read_csv("data/census_income_data.csv")
```

```python
a = df.drop("workclass", axis=1)
b = a.drop("fnlwgt", axis=1)
c = b.drop("education", axis=1)
d = c.drop("education.num", axis=1)
e = d.drop("marital.status", axis=1)
f = e.drop("occupation", axis=1)
g = f.drop("relationship", axis=1)
h = g.drop("race", axis=1)
i = h.drop("sex", axis=1)
j = i.drop("native.country", axis=1)
k = j.drop("income", axis=1)
l = k.drop("capital.gain", axis=1)
m = l.drop("capital.loss", axis=1)
```
