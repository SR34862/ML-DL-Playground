```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data/dm_office_sales.csv")
```

```python
sns.scatterplot(x="sales", y="salary", data=df, hue="division")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
x = df[["sales"]]
y = df[["salary"]]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.1, random_state=325)
```

```python
scaler = StandardScaler()
scaler.fit(x_train)
scaled_x_train = scaler.transform(x_train)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn_model = KNeighborsClassifier(n_neighbors=1)
```

```python
knn_model.fit(scaled_x_train, y_train)
```

```python
full_test = pd.concat([x_test, y_test], axis=1)
len(full_test)
```

```python
sns.scatterplot(x="sales", y="salary", data=full_test, alpha=0.7)
```
