```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data/books.csv")
```

```python
df.head()
```

```python
sns.boxplot(x=df["Genre"], y=df["Price"])
```

```python
sns.boxplot(x=df["Genre"], y=df["Reviews"])
```

```python
fiction_books = df[df["Genre"] == "Fiction"]
t5 = fiction_books.sort_values(by="Price",ascending=False).head(5)
print(t5)
```

```python
nf_books = df[df["Genre"] == "Non Fiction"]
nf_t5 = nf_books.sort_values(by="Price", ascending=False).head(5)
print(nf_t5)
```

```python
sns.boxplot(x=df["Genre"],y=df["UserRating"])
```

```python
df['Year'] = pd.to_numeric(df['Year'], errors='coerce')

filtered_df = df[(df['Year'] >= 2009) & (df['Year'] <= 2019)]

fig, axes = plt.subplots(2, 1, figsize=(12, 12))

sns.barplot(x='Year', y='Reviews', hue='Genre', data=filtered_df, estimator=sum, ax=axes[0])
axes[0].set_title('Number of Reviews from 2009 to 2019')

sns.barplot(x='Year', y='UserRating', hue='Genre', data=filtered_df, estimator=sum, ax=axes[1])
axes[1].set_title('Number of User Ratings from 2009 to 2019')
```
