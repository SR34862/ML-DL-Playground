```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data/AMES_Final_DF.csv")
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
sns.scatterplot(x=df["Gr Liv Area"], y=df["SalePrice"])
```

```python
from sklearn.model_selection import train_test_split
```

```python
x = df.drop("Gr Liv Area", axis=1)
y = df.drop("SalePrice", axis=1)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.1, random_state=325)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
scaled_x_train = scaler.fit_transform(x_train)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
scaled_x_train.shape
```

```python
l1 = LinearRegression()
l1.fit(scaled_x_train, y_train)
```

```python
x_test.shape
```

```python
y_pred = l1.predict(x_test)
```

```python
from sklearn.metrics import r2_score
```

```python
r2 = r2_score(y_pred, y_test)
```
