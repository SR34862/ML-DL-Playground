```python
reviews_df_5000 = reviews_df.iloc[:5000, :]
```

```python
def SIA_reviews(df):
  result = {}
  for i, row in tqdm(df.iterrows(), total=len(df)):
    text = row["Text"]
    my_id = row["Id"]
    result[my_id] = sia.polarity_scores(text)

  return result
```

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from scipy.special import softmax
```

```python
MODEL = f"cardiffnlp/twitter-roberta-base-sentiment-latest"
```

```python
tokenizer = AutoTokenizer.from_pretrained(MODEL)
model = AutoModelForSequenceClassification.from_pretrained(MODEL)
```

```python
encoded_text = tokenizer("I feel vibrant, energetic and confident", return_tensors="pt")
output1 = model(**encoded_text)
scores = output1[0][0].detach().numpy()
scores = softmax(scores)
display(scores)
```

```python
encoded_text = tokenizer("I feel ashamed, disgusted and cheated", return_tensors="pt")
output2 = model(**encoded_text)
scores1 = output2[0][0].detach().numpy()
scores1 = softmax(scores1)
display(scores1)
```

```python
encoded_text = tokenizer("Overall I feel alright and balanced", return_tensors="pt")
output3 = model(**encoded_text)
scores2 = output3[0][0].detach().numpy()
scores2 = softmax(scores2)
display(scores2)
```

```python
from tqdm.notebook import tqdm
```

```python
#2. Running Roberta Model for dataset
def polarity_scores_roberta(items:str)->dict:
  encoded_text = tokenizer(items, return_tensors="pt")
  output = model(**encoded_text)
  scores = output[0][0].detach().numpy()
  scores = softmax(scores)

  scores_dict = {
      "roberta_neg": scores[0],
      "roberta_neu": scores[1],
      "roberta_pos": scores[2]
  }

  return scores_dict
```

```python
reviews_df_5000 = reviews_df.iloc[:5000, :]
```

```python
result = {}

for index, row in tqdm(reviews_df_5000.iterrows(), total=len(reviews_df_5000)):
  text = row["Text"]
  my_id = row["Id"]
  try:
    #1st model - Vader with modified keys
    vader_result = sia.polarity_scores(text)
    newname = {}
    for key, value in vader_result.items():
      newname[f"vader_{key}"] = value
    #2nd model - For Roberta Model
    roberta_result = polarity_scores_roberta(text)
    both = newname | roberta_result
    result[my_id] = both
    #3rd model - For BERT Model
  except RuntimeError as _:
    print(f"Broke for index {index}")
```

```python
#Inserting the result into a dataframe
result_df = pd.DataFrame(result).T
result_df.reset_index(drop=True, inplace=True)
result_df.rename(columns={"index":"Id"}, inplace=True)
```
